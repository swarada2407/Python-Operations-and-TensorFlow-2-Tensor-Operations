```python

# Title: TensorFlow 2 Tensor Operations on a Dataset

import tensorflow as tf
import numpy as np
from sklearn.datasets import load_iris

# 1. Display library versions
print("=" * 60)
print("TENSORFLOW OPERATIONS ON IRIS DATASET")
print("=" * 60)

print("TensorFlow Version:", tf.__version__)
print("NumPy Version:", np.__version__)

# 2. Load the Iris dataset
iris = load_iris()

X = iris.data
y = iris.target

print("\n1. DATASET INFORMATION")
print("Dataset Name: Iris")
print("Number of Samples:", X.shape[0])
print("Number of Features:", X.shape[1])
print("Feature Names:", iris.feature_names)
print("Target Classes:", iris.target_names)

print("\nFirst 5 Dataset Rows:")
print(X[:5])

# 3. Convert dataset into TensorFlow tensors
X_tensor = tf.constant(X, dtype=tf.float32)
y_tensor = tf.constant(y, dtype=tf.int32)

print("\n2. TENSOR CREATION")
print("Feature Tensor:\n", X_tensor[:5].numpy())
print("Target Tensor:\n", y_tensor[:5].numpy())

# 4. Tensor properties
print("\n3. TENSOR PROPERTIES")
print("Shape:", X_tensor.shape)
print("Rank:", tf.rank(X_tensor).numpy())
print("Data Type:", X_tensor.dtype)

# 5. Basic Python operations
print("\n4. BASIC PYTHON OPERATIONS")
a = 10
b = 5

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

# 6. Tensor arithmetic on first 5 samples
print("\n5. TENSOR ARITHMETIC OPERATIONS")

sample = X_tensor[:5]

print("Original Data:\n", sample.numpy())
print("Addition (+ 2):\n", (sample + 2).numpy())
print("Subtraction (- 1):\n", (sample - 1).numpy())
print("Multiplication (* 2):\n", (sample * 2).numpy())
print("Division (/ 2):\n", (sample / 2).numpy())

# 7. Mathematical functions
print("\n6. MATHEMATICAL FUNCTIONS")

print("Square:\n", tf.square(sample).numpy())
print("Square Root:\n", tf.sqrt(sample).numpy())
print("Mean of Each Feature:\n",
      tf.reduce_mean(X_tensor, axis=0).numpy())
print("Maximum of Each Feature:\n",
      tf.reduce_max(X_tensor, axis=0).numpy())
print("Minimum of Each Feature:\n",
      tf.reduce_min(X_tensor, axis=0).numpy())

# 8. Reshape the dataset
print("\n7. RESHAPING TENSORS")

small_tensor = tf.reshape(X_tensor[:6], (3, 8))

print("Original Shape:", X_tensor[:6].shape)
print("Reshaped Shape:", small_tensor.shape)
print("Reshaped Tensor:\n", small_tensor.numpy())

# 9. NumPy conversion
print("\n8. TENSORFLOW TO NUMPY")

numpy_array = X_tensor.numpy()

print("TensorFlow Type:", type(X_tensor))
print("NumPy Type:", type(numpy_array))
print("NumPy Shape:", numpy_array.shape)
print("NumPy Mean:", np.mean(numpy_array, axis=0))

# 10. Create special tensors
print("\n9. SPECIAL TENSORS")

print("Zeros:\n", tf.zeros((2, 3)).numpy())
print("Ones:\n", tf.ones((2, 3)).numpy())

# 11. Matrix multiplication
print("\n10. MATRIX MULTIPLICATION")

matrix_a = tf.constant([[1.0, 2.0],
                        [3.0, 4.0]])

matrix_b = tf.constant([[5.0, 6.0],
                        [7.0, 8.0]])

print("Result:\n", tf.matmul(matrix_a, matrix_b).numpy())

# 12. Final output
print("\nDataset processing completed successfully!")
print("=" * 60)
```
